```python
import json
import os
import shutil

BASE = 'food101_dataset/food-101/food-101'
IMAGES = f'{BASE}/images'

TRAIN_DIR = 'food101_dataset/train'
TEST_DIR = 'food101_dataset/test'

os.makedirs(TRAIN_DIR, exist_ok=True)
os.makedirs(TEST_DIR, exist_ok=True)


with open(f'{BASE}/meta/train.json', 'r') as f:
    train_json = json.load(f)

for cls, img_list in train_json.items():
    dst_dir = f'{TRAIN_DIR}/{cls}'
    os.makedirs(dst_dir, exist_ok=True)
    for img_path in img_list:
        src = f'{IMAGES}/{img_path}.jpg'
        shutil.copy(src, dst_dir)

print("Train copy done")


with open(f'{BASE}/meta/test.json', 'r') as f:
    test_json = json.load(f)

for cls, img_list in test_json.items():
    dst_dir = f'{TEST_DIR}/{cls}'
    os.makedirs(dst_dir, exist_ok=True)
    for img_path in img_list:
        src = f'{IMAGES}/{img_path}.jpg'
        shutil.copy(src, dst_dir)

print("Test copy done")
```

image resize


```python
from PIL import Image
import os

IMG_SIZE = (224, 224)
TRAIN_DIR = 'food101_dataset/train'
TEST_DIR = 'food101_dataset/test'

def resize_all_images(folder_path):
    for class_folder in os.listdir(folder_path):
        class_path = f'{folder_path}/{class_folder}'
        for img_file in os.listdir(class_path):
            img_path = f'{class_path}/{img_file}'
            img = Image.open(img_path).convert('RGB')
            img = img.resize(IMG_SIZE)
            img.save(img_path)

resize_all_images(TRAIN_DIR)
print("Train resize done")

resize_all_images(TEST_DIR)
print("Test resize done")
```

```python
import tensorflow as tf

train_data = tf.keras.utils.image_dataset_from_directory(
    'food101_dataset/train',
    image_size=(224, 224),
    batch_size=32,
    label_mode='int'
)

test_data = tf.keras.utils.image_dataset_from_directory(
    'food101_dataset/test',
    image_size=(224, 224),
    batch_size=32,
    label_mode='int'
)


normalization_layer = tf.keras.layers.Rescaling(1./255)
train_data = train_data.map(lambda x, y: (normalization_layer(x), y))
test_data = test_data.map(lambda x, y: (normalization_layer(x), y))

print("Train batches:", len(train_data))
print("Test batches:", len(test_data))
```

```python
import tensorflow as tf


base_model = tf.keras.applications.ResNet50(
    weights='imagenet',
    include_top=False,
    input_shape=(224, 224, 3)
)
base_model.trainable = False


model = tf.keras.Sequential([
    base_model,
    tf.keras.layers.GlobalAveragePooling2D(),
    tf.keras.layers.Dense(256, activation='relu'),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(101, activation='softmax')
])


model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

```python
history = model.fit(
    train_data,
    validation_data=test_data,
    epochs=20
)
```

```python

loss, accuracy = model.evaluate(test_data)
print(f"Test Loss: {loss:.4f}")
print(f"Test Accuracy: {accuracy*100:.2f}%")
```

```python

train_loss, train_accuracy = model.evaluate(train_data)
print(f"Train Loss: {train_loss:.4f}")
print(f"Train Accuracy: {train_accuracy*100:.2f}%")
```

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Train')
plt.plot(history.history['val_accuracy'], label='Val')
plt.title('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Train')
plt.plot(history.history['val_loss'], label='Val')
plt.title('Loss')
plt.legend()

plt.show()
```
